```python
#Merge types
left_demo = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["A", "B", "C"]
})

right_demo = pd.DataFrame({
    "ID": [2, 3, 4],
    "Score": [80, 90, 70]
})

print("Inner:")
display(pd.merge(left_demo, right_demo, on="ID", how="inner"))

print("Left:")
display(pd.merge(left_demo, right_demo, on="ID", how="left"))

print("Outer:")
display(pd.merge(left_demo, right_demo, on="ID", how="outer"))
```

```python
# Outliers using IQR
scores = clean["Exam_Score"].dropna()

Q1 = scores.quantile(0.25)
Q3 = scores.quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = clean[
    (clean["Exam_Score"] < lower) |
    (clean["Exam_Score"] > upper)
]

print("Lower limit:", lower)
print("Upper limit:", upper)
print("Number of outliers:", len(outliers))
display(outliers[["Name", "Exam_Score"]])
```

```python
#apply()
def grade(score):
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 50:
        return "C"
    else:
        return "F"

apply_demo = clean[["Name", "Exam_Score"]].copy()
apply_demo["Grade"] = apply_demo["Exam_Score"].apply(grade)

display(apply_demo.head(10))
```

```python
#map()
status_map = {
    "Paid": 1,
    "Pending": 0
}

map_demo = clean[["Name", "Fee_Status"]].copy()
map_demo["Fee_Status_Code"] = map_demo["Fee_Status"].map(status_map)

display(map_demo.head(10))
```

```python
#omplete data-cleaning workflow
clean_df = df.copy()

# 1. Remove exact duplicates
clean_df = clean_df.drop_duplicates()

# 2. Clean text
clean_df["Name"] = clean_df["Name"].str.strip().str.title()
clean_df["Department"] = clean_df["Department"].str.strip().str.title()

# 3. Standardize categories
clean_df["Department"] = clean_df["Department"].replace({
    "Ai": "AI"
})

# 4. Convert types
clean_df["Enrollment_Date"] = pd.to_datetime(
    clean_df["Enrollment_Date"],
    errors="coerce"
)

clean_df["Attendance"] = pd.to_numeric(
    clean_df["Attendance"],
    errors="coerce"
)

clean_df["Study_Hours"] = pd.to_numeric(
    clean_df["Study_Hours"],
    errors="coerce"
)

# 5. Handle missing numerical values
for col in ["Attendance", "Study_Hours"]:
    clean_df[col] = clean_df[col].fillna(clean_df[col].median())

# 6. Handle missing categorical value
clean_df["Department"] = clean_df["Department"].fillna("Unknown")

# 7. Validate
print("Shape:", clean_df.shape)
print("\nMissing values:")
print(clean_df.isna().sum())
print("\nDuplicates:", clean_df.duplicated().sum())

display(clean_df.head())
```

```python
# Encoding categorical data
encoded = pd.get_dummies(
    clean_df,
    columns=["Gender", "Department", "City", "Fee_Status"],
    dtype=int
)

display(encoded.head())
print("Encoded shape:", encoded.shape)
```
