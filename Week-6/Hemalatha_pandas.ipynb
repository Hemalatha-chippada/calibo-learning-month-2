## Q1. Two NumPy arrays have shapes (3, 1) and (1, 3). Predict the shape of their element-wise sum and explain how broadcasting works. Why might the result differ from adding two one-dimensional arrays?

```python
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

result = a + b

print(result)
print(result.shape)
```

1)Predict the output of the given code.

2)Identify the shapes of arrays a and b before addition.

3)Explain how NumPy broadcasting enables the addition of these arrays.

4)What would happen if b were replaced with np.array([1, 2])? Explain the reason.

**Answer:**



### 1. Predict the output of the given code

```text
[[11 12 13]
 [21 22 23]
 [31 32 33]]
(3, 3)
```

### 2. Shapes of arrays before addition

* Shape of `a` = `(3, 1)`
* Shape of `b` = `(1, 3)`

### 3. How broadcasting enables the addition

NumPy broadcasting allows arrays with compatible shapes to be used in element-wise operations without manually changing their sizes.

Here, `a` has shape `(3, 1)` and `b` has shape `(1, 3)`. NumPy broadcasts them to a common shape of `(3, 3)` and performs element-wise addition.

For example:

```text
[10] + [1 2 3] = [11 12 13]
[20] + [1 2 3] = [21 22 23]
[30] + [1 2 3] = [31 32 33]
```

### 4. What happens if `b = np.array([1, 2])`?

The shapes become:

* `a` = `(3, 1)`
* `b` = `(2,)`

These shapes **are compatible for broadcasting**. NumPy treats `b` as having shape `(1, 2)` and broadcasts both arrays to `(3, 2)`.

The result will be:

```text
[[11 12]
 [21 22]
 [31 32]]
```

The result shape is:

```text
(3, 2)
```

This differs from adding two one-dimensional arrays because a 2D array with shape `(3, 1)` can broadcast across the columns, while two 1D arrays are aligned element-by-element and normally need compatible lengths.


# Code executed output

````python
import numpy as np

a = np.array([[10], [20], [30]])
b = np.array([[1, 2, 3]])

print("Shape of a:", a.shape)
print("Shape of b:", b.shape)

result = a + b

print(result)
print(result.shape)
````

### Explanation

Broadcasting means NumPy automatically adjusts the shapes of compatible arrays so that element-wise operations can be performed.

Here:

- `a` has shape `(3, 1)` → 3 rows and 1 column.
- `b` has shape `(1, 3)` → 1 row and 3 columns.

NumPy broadcasts both arrays to a common shape of `(3, 3)`:

```text
a:              b:

[10]            [1 2 3]
[20]            [1 2 3]
[30]            [1 2 3]

# Code executed output

````python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Salary": [25000, 30000, 35000, 10000000]
})

print("Mean:", df["Salary"].mean())
print("Median:", df["Salary"].median())
````

### Explanation

The **mean** is calculated by adding all salaries and dividing by the number of employees. Because `10,000,000` is extremely large compared with the other salaries, it greatly increases the mean to `2,522,500`.

The **median** is the middle value after arranging the salaries in ascending order. The salaries are:

`25000, 30000, 35000, 10000000`

The middle two values are `30000` and `35000`, so the median is `32,500`.

Therefore, the **median** better represents a typical employee's salary because it is less affected by the extreme value.

### Step 1 – Load the CSV file and inspect the dataset
`keep_default_na=False` is used on purpose so that the text `NA` is **kept as text** and can be investigated later (instead of silently becoming `NaN`). Only truly blank cells are treated as missing.

````python
import pandas as pd
import numpy as np

df = pd.read_csv("student_assessment_dirty.csv",
                 keep_default_na=False, na_values=[""])
rows_before = len(df)

print(df.head())
print()
df.info()
print()
print(df.isnull().sum())
````

### Step 2 – Standardize department names

````python
df["Department"] = (df["Department"].str.strip()
                                    .str.upper()
                                    .str.replace(".", "", regex=False))
df["Department"] = df["Department"].replace({"FINANCE": "Finance"})

print(df["Department"].value_counts())
````

```

**Explanation:** `IT`, `it` and `I.T.` all become `IT` (uppercase and dots removed), and `finance` / `Finance` both become `Finance`. This count still includes the duplicate row, which is removed in the next step. Standardizing *before* removing duplicates makes sure rows that differ only in spelling are matched correctly.

### Step 3 – Identify and remove exact duplicate records

````python
print("Duplicate rows found:", df.duplicated().sum())
print(df[df.duplicated(keep=False)])

df = df.drop_duplicates()
records_removed = rows_before - len(df)

print("Records removed:", records_removed)
print("Rows remaining:", len(df))
````

**Explanation:** `duplicated()` finds identical rows and `drop_duplicates()` keeps the first copy. The repeated S102 (Ravi) record is removed.

### Step 4 – Convert Marks to a numeric format
The original text is kept in `Marks_Raw` so that missing values can be investigated in Step 5.

````python
df["Marks_Raw"] = df["Marks"]

df["Marks"] = pd.to_numeric(
    df["Marks"].str.extract(r"(\d+\.?\d*)")[0],
    errors="coerce"
)

print(df[["Student_ID", "Student_Name", "Marks_Raw", "Marks"]])
print()
print(df["Marks"].dtype)
````

### Step 5 – Investigate missing marks (absent vs. not entered)

````python
df["Marks_Status"] = "Entered"
df.loc[df["Marks_Raw"].isna(), "Marks_Status"] = "Not entered (blank)"
df.loc[df["Marks_Raw"] == "NA", "Marks_Status"] = "NA - check source"
df.loc[df["Marks_Raw"].str.strip().str.lower() == "absent", "Marks_Status"] = "Absent"

missing = df[df["Marks"].isna()]
print(missing[["Student_ID", "Student_Name", "Marks_Raw", "Marks_Status"]])
````

- **Sneha** has `NA`, which is unclear (not applicable? not entered?) and must be checked with the source.
