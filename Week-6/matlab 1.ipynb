```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
print("Matplotlib version:", plt.matplotlib.__version__)
print("Pandas version:", pd.__version__)
print("Numpy version:", pd.__version__)

```

```python
#matplotlib object version
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.figure(figsize=(6, 4))
plt.plot(x, y)
plt.title("Object-oriented interface: Figure + pltes")
plt.xlabel("x")
plt.ylabel("cos(x)")
plt.show()
print("Type of plt:", type(plt))
```

```python
#pyplot interface -- same plot, different style of writing it
plt.figure(figsize=(6, 4))
plt.plot(x, y)
plt.title("pyplot interface (same result)")
plt.xlabel("x")
plt.ylabel("cos(x)")
plt.show()
```

```python
df = pd.read_csv("C:\\Users\\chipp\\Downloads\\web_traffic_daily.csv")
df.head()
```

```python
plt.Figure(figsize=(10, 4))
plt.plot(df["date"], df["visits"], color="tab:blue", linewidth=1.5, label="visits")
plt.title("Daily visits over one year")
plt.xlabel("Date")
plt.ylabel("Visits")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python

plt.Figure(figsize=(10, 4))
plt.plot(df["date"], df["visits"],  color="tab:blue", linewidth=1.5, label="visits")
plt.plot(df["date"], df["signups"], color="tab:pink", linewidth=1.5, linestyle="--", label="signups")
plt.title("Daily visits over one year")
plt.xlabel("Date")
plt.ylabel("Visits")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
#why dates are overlapping
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(df["date"], df["visits"],  color="tab:blue", linewidth=1.5, label="visits", linestyle="--")
ax.plot(df["date"], df["signups"], color="tab:pink", linewidth=1.5, linestyle="--", label="signups")
ax.set_title("Daily visits over one year")
ax.set_xlabel("Date")
ax.set_ylabel("Visits")
ax.legend()
ax.set_yscale("log")

```

```python
#scatter plot
plt.figure(figsize=(6, 5))
sc=plt.scatter(df["page_load_ms"], df["error_count"],
               c=df["visits"], cmap="viridis", alpha=0.7)
plt.title("Errors vs page load time (colour = visits)")
plt.xlabel("page load time (ms)")
plt.ylabel("Error count")
plt.colorbar(sc, label="Visits")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sc=plt.scatter(df["page_load_ms"], df["error_count"],
               c=df["deployment_flag"], cmap="coolwarm", alpha=0.7)
plt.title("Errors vs page load time (colour = deployment_flag)")
plt.xlabel("page load time (ms)")
plt.ylabel("Error count")
plt.colorbar(sc, label="deployment_flag")
plt.show()
```

```python
#bar chart
df["date"] = pd.to_datetime(df["date"], dayfirst=True)
monthly = df.set_index("date") .resample("ME") [["visits", "signups"]].sum()
monthly.index = monthly.index.strftime("%b")
plt.figure(figsize=(10, 4))
x = np.arange(len(monthly))
width = 0.35
plt.bar(
    x - width/2,
    monthly["visits"],
    width,
    label="Visits"
)
plt.bar( x + width/2,monthly["signups"] * 10, width, label = "signups (x10 for scale)")
plt.xticks(x, monthly.index, rotation=45)
plt.title("Monthly Visits vs Signups (Grouped Bar)")
plt.xlabel("Month")
plt.ylabel("Count")
plt.legend()
plt.grid(axis="y", alpha=0.3)
```
