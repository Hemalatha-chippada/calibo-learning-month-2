```python
#json arrays
student_json = '''
[
      {"id": 1, "name": "Ravi", "marks": 82},
      {"id": 2, "name": "Priya", "marks": 91},
      {"id": 3, "name": "sai", "marks": 76}
]
'''
students = json.loads(student_json)
print("Number of students:", len(students))
for student in students:
    print(student["name"], "scored", student["marks"])
top_students = [s for s in students if s["marks"] >= 80]
print("students scoring 80 or more:", [s["name"] for s in top_students])    
```

```python
#handling invalid json
import json
invalid_json = "{'name': 'Ravi', 'age': 22}"
try:
    data = json.loads(invalid_json)
    print(data)
except json.JSONDecodeError as error:
    print("Could not parse JSON.")
    print("Reasn:", error)       
```

```python
import json
valid_json = '{"name": "Ravi", "age": 22}'
try:
    data = json.loads(valid_json)
    print(data)
except json.JSONDecodeError as error:
    print("Invalid JSON:", error)
```

```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel

app = FastAPI(title="Campus Cafe")

orders = {1: {"student": "Asha", "item": "Cold coffee", "qty": 2,
              "price": 60.5, "extras": ["ice"], "paid": False}}

class Order(BaseModel):          # validates the incoming JSON
    student: str
    item: str
    qty: int
    price: float
    extras: list[str] = []
    paid: bool = False

@app.get("/orders")
def all_orders():
    return orders

@app.get("/orders/{oid}")
def one_order(oid: int):
    if oid not in orders:
        raise HTTPException(404, "Order not found")
    return orders[oid]

@app.post("/orders", status_code=201)
def new_order(o: Order):
    oid = max(orders, default=0) + 1
    orders[oid] = o.model_dump()
    return {"id": oid, **orders[oid]}

@app.put("/orders/{oid}")
def replace_order(oid: int, o: Order):
    if oid not in orders:
        raise HTTPException(404, "Order not found")
    orders[oid] = o.model_dump()
    return orders[oid]

@app.delete("/orders/{oid}")
def cancel_order(oid: int):
    if orders.pop(oid, None) is None:
        raise HTTPException(404, "Order not found")
    return {"cancelled": oid}
```
