C1. Given a 1D NumPy array of exam scores, use np.where() to produce labels 'P' (score >= 40) or 'F' (score < 40).

The given 1D NumPy array contains the exam scores of different students. The np.where() function checks each score against the condition scores >= 40. If the condition is true, the student is given the label P, which means Pass. If the condition is false, the student is given the label F, which means Fail. Therefore, scores below 40 are labelled F, while scores of 40 or above are labelled P.

C2. Write a function standardize(x) that takes a 1D NumPy array and returns its z-score standardized version using only NumPy. 

**Answer**


````python
import numpy as np

def standardize(x):
    mean = np.mean(x)
    std = np.std(x)
    return (x - mean) / std

x = np.array([10, 20, 30, 40, 50])

result = standardize(x)

print(result)
````

Explanation

The standardize() function takes a 1D NumPy array as input and calculates its mean and standard deviation using NumPy. It then subtracts the mean from each value and divides the result by the standard deviation. This converts the values into their z-score standardized form, where the values are centered around a mean of 0.

C7. Given a grayscale image as a 2D NumPy array with pixel values in [0, 255], normalize the values to [0, 1] using min-max normalization. 

**Answer**


````python
import numpy as np

image = np.array([[0, 50, 100],
                  [150, 200, 255]])

min_value = np.min(image)
max_value = np.max(image)

normalized_image = (image - min_value) / (max_value - min_value)

print("Original image:")
print(image)

print("Normalized image:")
print(normalized_image)
````

You are given a raw 1D or 2D NumPy array containing NaN values.

(a) detect missing values

(b) handle them appropriately and justify your strategy, and

**Answer**

````python
import numpy as np

data = np.array([10, 20, np.nan, 40, 50])

mean_value = np.nanmean(data)
clean_data = np.where(np.isnan(data), mean_value, data)

print("\n Mean used to replace NaN:")
print(mean_value)

print("Clean data:")
print(clean_data)
````

Explanation

Next, the missing value needs to be handled before using the data for a model. Here, np.nanmean(data) calculates the mean while ignoring the NaN value. The np.where() function then replaces the missing value with this mean. This strategy keeps the same number of data values without deleting any sample.

(c) normalize or standardize the clean data before it is fed to a model.

**Answer**

````python
mean = np.mean(clean_data)
std = np.std(clean_data)

standardized_data = (clean_data - mean) / std

print("\n(c) Standardized data:")
print(standardized_data)
````

Explanation

After handling the missing value, the cleaned data is standardized. We calculate its mean and standard deviation using np.mean() and np.std(). The formula (data - mean) / std converts the values to a common scale, with the resulting data centered around 0.

Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.


````python
import numpy as np

# Sample array containing a NaN value
data = np.array([10, 20, np.nan, 40, 50])

# (a) Detect missing values
missing = np.isnan(data)

print("(a) Missing values:")
print(missing)

# (b) Handle missing value using the mean
mean_value = np.nanmean(data)
clean_data = np.where(np.isnan(data), mean_value, data)

print("\n(b) Clean data:")
print(clean_data)

# (c) Standardize the clean data
mean = np.mean(clean_data)
std = np.std(clean_data)

standardized_data = (clean_data - mean) / std

print("\n(c) Standardized data:")
print(standardized_data)
````

Final Result

The final data is now clean, contains no NaN values, and is standardized to a common scale. This cleaned and standardized data can now be provided as input to a machine-learning model.

E1. Per-Column Min–Max Normalization 

Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.

**Answer**



````python
import numpy as np

X = np.array([[10, 100, 1000],
              [20, 200, 2000],
              [30, 300, 3000],
              [40, 400, 4000]])

# Find minimum and maximum of each column
col_min = np.min(X, axis=0)
col_max = np.max(X, axis=0)

# Per-column Min-Max normalization
X_normalized = (X - col_min) / (col_max - col_min)

print("Original matrix:")
print(X)

print("\nColumn-wise minimum:")
print(col_min)

print("\nColumn-wise maximum:")
print(col_max)

print("\nNormalized matrix:")
print(X_normalized)
````

Explanation

The matrix contains multiple features as columns, so np.min(X, axis=0) and np.max(X, axis=0) find the minimum and maximum separately for each column. The formula (X - col_min) / (col_max - col_min) is then applied using NumPy broadcasting, so there is no need for explicit loops. Each column is independently scaled to the range 0 to 1. This is different from using one global minimum and maximum for the entire matrix.

E2. Three-Level Score Classification 


Given a 1D array of exam scores, use a single vectorized expression (no if/else statements and no explicit loops) to classify each score as:


Fail (< 40), Average (40–70 inclusive), or Excellent (> 70).


Hint: You may need to nest or combine np.where() calls.

**Answer**


````python
import numpy as np

scores = np.array([25, 40, 55, 70, 75, 90])

result = np.where(scores < 40, "Fail",
                  np.where(scores <= 70, "Average", "Excellent"))

print("Scores:", scores)
print("Classification:", result)
````

The first np.where() checks whether each score is less than 40; if true, it assigns "Fail". For scores that are not below 40, the second np.where() checks whether the score is less than or equal to 70 and assigns "Average". Scores greater than 70 are classified as "Excellent".
